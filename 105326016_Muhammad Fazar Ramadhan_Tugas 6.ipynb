```python
skor_nobita = 0
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif 1 <= jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        return 10
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif 1 <= jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        return 20
    else:
        print("Nobita tidur terlalu lama")
        return -20
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50
def jalankan_aktivitas():
    print("\n=== MENU AKTIVITAS ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main Bersama Doraemon")
    print("4. Menyontek saat ujian")
    
    pilihan = int(input("Masukkan pilihan aktivitas: "))
    
    if pilihan == 1:
        jam = float(input("Berapa jam Nobita belajar hari ini? "))
        return belajar(jam)
    
    elif pilihan == 2:
        jam = float(input("Berapa jam Nobita tidur siang hari ini? "))
        return tidur(jam)
    
    elif pilihan == 3:
        g = int(input("Berapa gadget yang digunakan Nobita? "))
        w = float(input("Berapa jam Nobita bermain gadget? "))
        return main_bersama_doraemon(g, w)
    
    elif pilihan == 4:
        return menyontek()
    
    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
        return 0
def evaluasi_akhir(skor):
    if skor >= 80:
        return "Siswa Teladan"
    elif skor >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"
def menu_utama():
    global skor_nobita
    print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
    
    lanjut = "y"
    while lanjut.lower() == "y":
        nilai = jalankan_aktivitas()
        update_skor(nilai)

        print(f"Skor sementara Nobita: {skor_nobita}")
        lanjut = input("Ingin memasukkan aktivitas lain? (y/n): ")
    
    print("\n=== HASIL AKHIR ===")
    print("Total skor:", skor_nobita)
    print("Kategori:", evaluasi_akhir(skor_nobita))
menu_utama()
```
