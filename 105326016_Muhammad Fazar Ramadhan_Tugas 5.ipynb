```python
# 3. membantu belajar mengatur keuangan Nobita
# a. pengeluaran harian Nobita
hari = 1
saldo = 50000
no = 1
# Perulangan berjalan selama hari <= 30
while hari <= 30:
    pengeluaran = int(input(f'{no}. Pengeluaran hari ke-{hari}:'))
    print(f"{no}. Pengeluaran hari ke-{hari} : Rp.{pengeluaran:,.0f}")
    no += 1
    saldo -= pengeluaran


    
    # Cek apakah saldo menjadi minus (< 0)
    if saldo < 0:
        print(f"{no}. Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break

    # Jika saldo masih mencukupi (>= 0), tampilkan sisa saldo lalu tambah nilai hari
    else:
        print(f"{no}. Sisa saldo hari ke-{hari}: Rp.{saldo:,.0f}")
        no += 1
        hari += 1
```

```python
# b. menganalisis pengeluaran mingguan Nobita
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{i+1}: Nobita tidak jajan, dilewati analisisnya.")
        continue
    print(f"Minggu ke-{i+1}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis mingguan selesai.")
```

```python
# c. Menyusun laporan bulanan Nobita

# Data pengeluaran per minggu untuk 3 bulan sesuai contoh soal
data_bulan = [
    [5000, 10000, 15000, 5000],   # Bulan ke-1
    [2000, 1000, 1000, 40000],    # Bulan ke-2 (minggu ke-4 > 20000)
    [70000, 5000, 5000, 5000]     # Bulan ke-3 (minggu ke-1 > 20000)
]

for bulan in range(1, 4):
    print(f"Bulan ke-{bulan}")
    total_bulanan = 0
    
    for minggu in range(1, 5):
        pengeluaran = data_bulan[bulan-1][minggu-1]
        
        # Cek apakah pengeluaran > 20000
        if pengeluaran > 20000:
            print(f"Pengeluaran minggu ke-{minggu}: {pengeluaran}")
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            break
            
        print(f"Pengeluaran minggu ke-{minggu}: {pengeluaran}")
        total_bulanan += pengeluaran
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")
    
    print() # Baris kosong antar bulan
```

```python
# d. Evaluasi Akhir Tahun 
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print(f"=== Bulan ke-{bulan} ===")
    minggu = 1
    
    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")
        
        if libur.lower() == 'y':
            print("Minggu ini dilewati (continue).")
            minggu += 1
            continue
        elif libur.lower() == 'n':
            pengeluaran = int(input("Masukkan pengeluaran minggu ini: "))
            saldo -= pengeluaran
            
            if saldo < 0:
                break
        
        minggu += 1
    else:
        print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}")
        print()
        continue
    
    break
else:
    print(f"Nobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")
```
