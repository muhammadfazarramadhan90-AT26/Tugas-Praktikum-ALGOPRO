```python
# interest.py

def compound_interest(principal, rate, time):
    """
    Menghitung nilai akhir dengan bunga majemuk.
    Rumus: Future value = principal * (1 + rate) ** time
    """
    return principal * ((1 + rate) ** time)

def simple_interest(principal, rate, time):
    """
    Menghitung nilai total akhir (modal awal + total bunga sederhana).
    Rumus umum total akhir = principal + (principal * rate * time)
    Atau sesuai rumus dasar di modul: principal * rate * time (sebagai total bunga)
    Jika nilai akhir total: principal + (principal * rate * time)
    """
    # Berdasarkan contoh output di soal: 6000000 dengan rate 0.06 (6%) selama 10 tahun
    # Hasil sederhana = 9,600,000.00 (artinya modal + bunga sederhana)
    # 6000000 + (6000000 * 0.06 * 10) = 6000000 + 3600000 = 9600000.
    interest_earned = principal * rate * time
    return principal + interest_earned
```

```python
# display.py

def print_result(principal, rate, years, ci, si):
    """
    Berfungsi untuk menampilkan hasil perhitungan saldo awal, 
    bunga tahunan, lama menabung, nilai akhir bunga majemuk, 
    dan nilai akhir bunga sederhana.
    """
    print("\n===== HASIL PERHITUNGAN TABUNGAN =====")
    print(f"Saldo Awal        : Y{principal:,.1f}")
    print(f"Bunga Tahunan     : {rate * 100:.1f}%")
    print(f"Lama Menabung     : {years} tahun")
    print("-" * 45)
    print(f"Nilai Akhir (Bunga Majemuk)  : Y{ci:,.2f}")
    print(f"Nilai Akhir (Bunga Sederhana): Y{si:,.2f}")
    print("=" * 45)
```

```python
# main.py

import interest
import display

def main():
    print("=== Program Kalkulator Bunga Tabungan ===")
    
    # Menerima input dari pengguna
    principal = float(input("Masukkan saldo awal (contoh: 80000): "))
    rate = float(input("Masukkan bunga tahunan (misal 5% -> 0.05): "))
    years = int(input("Masukkan lama menabung (tahun): "))
    
    # Menghitung menggunakan fungsi dari modul interest
    ci_result = interest.compound_interest(principal, rate, years)
    si_result = interest.simple_interest(principal, rate, years)
    
    # Menampilkan hasil menggunakan fungsi dari modul display
    display.print_result(principal, rate, years, ci_result, si_result)

if __name__ == "__main__":
    main()
```
