```python
# Tugas 3 Praktikum Alpro
# 1. Membuat Data Belanjaan

    # List nama barang
belanjaan = ["Apel", "Susu", "Roti"]

    # Tuple harga dan jumlah barang
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

    # Tampilkan karakter pertama dari barang pertama dan karakter terakhir dari barang terakhir
print(f"Karakter pertama dari barang pertama: {belanjaan[0][0]}")
print(f"Karakter terakhir dari barang terakhir: {belanjaan[-1][-1]}")
print("-" * 40)

# 2. Hitung Total Harga per Barang dan Total Keseluruhan

    # Total per barang
total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]

    # Total keseluruhan
total_belanja = total_apel + total_susu + total_roti

    # Cetak hasil perhitungan
print(f"Total harga Apel  : Rp{total_apel:,}")
print(f"Total harga Susu  : Rp{total_susu:,}")
print(f"Total harga Roti  : Rp{total_roti:,}")
print(f"Total belanja keseluruhan: Rp{total_belanja:,}")

# 3. Cetak Struk Belanja

    # Data pembeli dan toko
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"

# Cetak struk dengan f-string dan escape sequence
print("=" * 40)
print(f"{nama_toko:^40}")
print("=" * 40)
print(f"Nama Pembeli : {nama_pembeli}")
print("-" * 40)
print(f"{'Barang':<10}{'Harga':>10}{'\tJumlah':>8}{'Total\t\t':>12}")
print("-" * 40)
print(f"{belanjaan[0]:<10}{harga_jumlah[0]:>10,}{harga_jumlah[1]:>8}{total_apel:>12,}")
print(f"{belanjaan[1]:<10}{harga_jumlah[2]:>10,}{harga_jumlah[3]:>8}{total_susu:>12,}")
print(f"{belanjaan[2]:<10}{harga_jumlah[4]:>10,}{harga_jumlah[5]:>8}{total_roti:>12,}")
print("-" * 40)
print(f"{'TOTAL':<28}Rp{total_belanja:>10,}")
print("=" * 40)
print("     Terima kasih telah belanja!🙏🏻\t")
print("=" * 40)

# 4. Tambahkan Barang Baru ke List

    # Tambahkan Pisang ke list belanjaan
belanjaan.append("Pisang")

    # Tampilkan list belanjaan terbaru
print(f"\nDaftar belanjaan setelah ditambah Pisang: {belanjaan}")
```
