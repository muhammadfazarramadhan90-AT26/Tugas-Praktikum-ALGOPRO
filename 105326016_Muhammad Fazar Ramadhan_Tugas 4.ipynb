```python
# Smart Vehicle Insurance Analyzer by Fazar

usia_pengemudi = int(input("Usia pengemudi: "))
jenis_kendaraan = input("Jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (Rp): "))
area_penggunaan = input("Area penggunaan (kota/desa): ").lower()
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? (ya/tidak): ").lower()

# 1: Risiko awal berdasarkan usia
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = "tinggi"
else:
    risiko = "rendah"

# 2: Risiko berdasarkan jenis kendaraan
if jenis_kendaraan == "motor":
    risiko_kendaraan = "sedang"
else:
    risiko_kendaraan = "rendah"

if risiko == "tinggi" or risiko_kendaraan == "tinggi":
    risiko_akhir = "tinggi"
elif risiko == "rendah" and risiko_kendaraan == "sedang":
    risiko_akhir = "sedang"
else:
    risiko_akhir = risiko

# 3: Area penggunaan
if area_penggunaan == "perkotaan":
    if risiko_akhir == "rendah":
        risiko_akhir = "sedang"
    elif risiko_akhir == "sedang":
        risiko_akhir = "tinggi"

# 4: Sistem pengaman
if sistem_pengaman == "tidak":
    if risiko_akhir == "rendah":
        risiko_akhir = "sedang"
    elif risiko_akhir == "sedang":
        risiko_akhir = "tinggi"

# 5: Premi asuransi
if risiko_akhir == "tinggi":
    premi = 0.1 * nilai_kendaraan
elif risiko_akhir == "sedang":
    premi = 0.07 * nilai_kendaraan
else:
    premi = 0.05 * nilai_kendaraan

# 6: Pesan Doraemon (conditional expression)
pesan = "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!" \
    if risiko_akhir == "tinggi" else "Risiko kendaraan masih aman menurut Doraemon."

# Output hasil
print("\n===== HASIL ANALISIS DORABIKE =====")
print(f"Kategori Risiko        : {risiko_akhir.capitalize()}")
print(f"Premi Asuransi         : Rp{premi:,.0f}")
print(f"Pesan Doraemon         : {pesan}")
```
